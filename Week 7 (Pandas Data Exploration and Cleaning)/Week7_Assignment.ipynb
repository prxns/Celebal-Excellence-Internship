# Week 7 Pandas Assignment

## Import Pandas

```python
import pandas as pd
```

## Step 1: Load CSV dataset

```python
df = pd.read_csv("Sample - Superstore.csv", encoding="latin1")
```

## Step 2: Explore the dataset

```python
# 1. Head
print("First 5 Rows")
print(df.head())

# 2. Tail
print("Last 5 Rows")
print(df.tail())

# 3. Shape
print("Shape")
print(df.shape)

# 4. Columns
print("Columns")
print(df.columns)

# 5. Data Types
print("Data Types")
print(df.dtypes)
```

## Step 3: Handle Missing Values

```python
print(df.isnull().sum())

df.fillna(0, inplace=True)
```

## Step 4: Basic Operations

```python
# Filter Rows
high_sales = df[df["Sales"] > 500]

high_sales.head()

# Select Columns
selected = df[["Order ID","Customer Name","Sales","Quantity"]]

selected.head()
```

## Step 5: Remove Duplicates

```python
print("Before:", len(df))

df = df.drop_duplicates()

print("After:", len(df))
```

## Step 6: Create a Derived Column

```python
df["total_amount"] = df["Sales"] * df["Quantity"]

df.head()
```

## Step 7: Save Cleaned Dataset

```python
df.to_csv("cleaned_superstore.csv", index=False)

print("Cleaned dataset saved successfully!")
```
